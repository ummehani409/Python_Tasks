```python
expenses_list = []

total_expenses = int(input("Total kitne expenses enter karne hain? "))
for i in range(total_expenses):
    print(f"\n--- Expense #{i + 1} ---")
    name = input("Expense Name enter karein: ")
    category = input("Category (Food/Transport/Shopping/Bills/Other): ")
    amount = float(input("Amount enter karein: "))
    expense_dict = {
        "name": name,
        "category": category,
        "amount": amount
    }
    expenses_list.append(expense_dict)

total_spending = 0
for expense in expenses_list:
    total_spending += expense["amount"]

category_spending = {}
for expense in expenses_list:
    category = expense["category"]
    amount = expense["amount"]
    if category in category_spending:
        category_spending[category] += amount
    else:
        category_spending[category] = amount

highest_expense = max(expenses_list, key=lambda expense: expense["amount"])
budget = float(input("\nApna budget enter karein: "))

if total_spending > budget:
    print("\nWARNING: Aapka budget exceed ho gaya hai!")
else:
    print("\nAapka budget exceed nahi hua.")

def final_report():
    print("\n========== FINAL REPORT ==========")
    print("Total Spending:", total_spending)
    print("Budget:", budget)
    print("\n--- Category Wise Spending ---")
    for category, amount in category_spending.items():
        print(category, ":", amount)
    print("\n--- Highest Expense ---")
    print("Name:", highest_expense["name"])
    print("Category:", highest_expense["category"])
    print("Amount:", highest_expense["amount"])

final_report()

products = {}

def add_product():
    name = input("Product Name: ")
    price = float(input("Price: "))
    quantity = int(input("Quantity: "))
    products[name] = {
        "price": price,
        "quantity": quantity
    }
    print("Product added successfully.")

def sell_product():
    name = input("Product Name: ")
    quantity = int(input("Quantity to sell: "))
    if name in products:
        if products[name]["quantity"] >= quantity:
            products[name]["quantity"] -= quantity
            print("Product sold successfully.")
        else:
            print("WARNING: Insufficient stock.")
    else:
        print("Product not found.")

def restock_product():
    name = input("Product Name: ")
    quantity = int(input("Quantity to add: "))
    if name in products:
        products[name]["quantity"] += quantity
        print("Product restocked successfully.")
    else:
        print("Product not found.")

def search_product():
    name = input("Product Name: ")
    if name in products:
        print("Product:", name)
        print("Price:", products[name]["price"])
        print("Quantity:", products[name]["quantity"])
    else:
        print("Product not found.")

def display_inventory():
    print("\n--- Inventory ---")
    total_value = 0
    for name, product in products.items():
        price = product["price"]
        quantity = product["quantity"]
        value = price * quantity
        print(name, "| Price:", price, "| Quantity:", quantity, "| Value:", value)
        total_value += value
    print("Total Inventory Value:", total_value)
    print("\n--- Low Stock Products ---")
    for name, product in products.items():
        if product["quantity"] < 5:
            print(name, "| Quantity:", product["quantity"])

while True:
    print("\n1. Add Product")
    print("2. Sell Product")
    print("3. Restock Product")
    print("4. Search Product")
    print("5. Display Inventory")
    print("6. Exit")
    choice = input("Enter your choice: ")
    if choice == "1":
        add_product()
    elif choice == "2":
        sell_product()
    elif choice == "3":
        restock_product()
    elif choice == "4":
        search_product()
    elif choice == "5":
        display_inventory()
    elif choice == "6":
        print("Program ended.")
        break
    else:
        print("Invalid choice.")

books = []

def add_book():
    book_id = input("Book ID: ")
    for book in books:
        if book["id"] == book_id:
            print("Book ID already exists.")
            return
    title = input("Book Title: ")
    author = input("Author: ")
    book = {
        "id": book_id,
        "title": title,
        "author": author,
        "available": True
    }
    books.append(book)
    print("Book added successfully.")

def display_books():
    print("\n--- All Books ---")
    for book in books:
        print("Book ID:", book["id"])
        print("Title:", book["title"])
        print("Author:", book["author"])
        print("Status:", "Available" if book["available"] else "Borrowed")
        print()

def search_book():
    title = input("Enter Book Title: ")
    for book in books:
        if book["title"].lower() == title.lower():
            print("Book ID:", book["id"])
            print("Title:", book["title"])
            print("Author:", book["author"])
            print("Status:", "Available" if book["available" else "Borrowed")
            return
    print("Book not found.")

def borrow_book():
    book_id = input("Enter Book ID: ")
    for book in books:
        if book["id"] == book_id:
            if book["available"]:
                book["available"] = False
                print("Book borrowed successfully.")
            else:
                print("Book is already borrowed.")
            return
    print("Book not found.")

def return_book():
    book_id = input("Enter Book ID: ")
    for book in books:
        if book["id"] == book_id:
            if not book["available"]:
                book["available"] = True
                print("Book returned successfully.")
            else:
                print("Book is already available.")
            return
    print("Book not found.")

def main():
    while True:
        print("\n1. Add Book")
        print("2. Display Books")
        print("3. Search Book")
        print("4. Borrow Book")
        print("5. Return Book")
        print("6. Exit")
        choice = input("Enter your choice: ")
        if choice == "1":
            add_book()
        elif choice == "2":
            display_books()
        elif choice == "3":
            search_book()
        elif choice == "4":
            borrow_book()
        elif choice == "5":
            return_book()
        elif choice == "6":
            print("Program ended.")
            break
        else:
            print("Invalid choice.")

main()

menu = {
    "Burger": 500,
    "Pizza": 1200,
    "Pasta": 800,
    "Drink": 200
}

orders = []
print("========= Restaurant Menu ========")
for key, value in menu.items():
    print(f'{key} : {value}')

while True:
    item_name = input("Enter your item name: ").title()
    if item_name in menu:
        quantity = int(input("Enter quantity of your item: "))
        price = menu[item_name]
        sub_total = price * quantity
        orders.append({
            "item": item_name,
            "quantity": quantity,
            "sub_total": sub_total
        })
    else:
        print("This item is not available")
    choice = input("You want to add more items: yes/no: ").lower()
    if choice == "no":
        break

total = sum(order["sub_total"] for order in orders)
tax = total * 0.13
discount = total * 0.1 if total >= 3000 else 0
grand_total = total + tax - discount

def receipt():
    print("\n============ Receipt ============")
    for order in orders:
        print(f'{order["item"]} x {order["quantity"]} = {order["sub_total"]}')
    print("Total: ", total)
    print("Tax: ", tax)
    print("Discount: ", discount)
    print("Grand Total: ", grand_total)

receipt()

patients = []

def add_patient():
    patient_id = int(input("Enter your id: "))
    for patient in patients:
        if patient["id"] == patient_id:
            print("Patient already exists")
            return
    patient_name = input("Enter patient name: ")
    patient_age = int(input("Enter patient age: "))
    problem = input("Enter patient problem: ")
    time = int(input("Enter appointment time: "))
    emergency = input("Is emergency patient? yes/no: ").lower()
    patients.append({
        "id": patient_id,
        "name": patient_name,
        "age": patient_age,
        "problem": problem,
        "time": time,
        "emergency": emergency
    })

def search_patient():
    patient_id = int(input("Enter patient id: "))
    for patient in patients:
        if patient["id"] == patient_id:
            print("Patient found")
            print(f'Patient name: {patient["name"]}')
            print(f'Patient age: {patient["age"]}')
            print(f'Patient problem: {patient["problem"]}')
            print(f'Patient time: {patient["time"]}')
            print(f'Patient emergency: {patient["emergency"]}')
            return
    print("Patient not found")

def display_appointment():
    print("========== Emergency patient ==========")
    for patient in patients:
        if patient["emergency"] == "yes":
            print(f'Patient name: {patient["name"]}')
            print(f'Patient age: {patient["age"]}')
            print(f'Patient problem: {patient["problem"]}')
            print(f'Patient time: {patient["time"]}')
    print("========== Normal patient ==========")
    for patient in patients:
        if patient["emergency"] == "no":
            print(f'Patient name: {patient["name"]}')
            print(f'Patient age: {patient["age"]}')
            print(f'Patient problem: {patient["problem"]}')
            print(f'Patient time: {patient["time"]}')

def delete_patient():
    patient_id = int(input("Enter patient id: "))
    for patient in patients:
        if patient["id"] == patient_id:
            patients.remove(patient)
            print("Patient deleted")
            return
    print("Patient not found")

def count_patient():
    emergency_count = sum(1 for p in patients if p["emergency"] == "yes")
    normal_count = len(patients) - emergency_count
    print(f"Emergency Patients: {emergency_count}, Normal Patients: {normal_count}")

while True:
    print("\n1. Add patient")
    print("2. Search patient")
    print("3. Display appointment")
    print("4. Delete patient")
    print("5. Count patient")
    print("6. Exit")
    try:
        choice = int(input("Enter your choice: "))
        if choice == 1:
            add_patient()
        elif choice == 2:
            search_patient()
        elif choice == 3:
            display_appointment()
        elif choice == 4:
            delete_patient()
        elif choice == 5:
            count_patient()
        elif choice == 6:
            break
        else:
            print("Invalid choice")
    except ValueError:
        print("Invalid choice, please enter a number")
```
