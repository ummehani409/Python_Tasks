```python
#1 even or odd
number = int(input("Enter a Number: "))
if number % 2 == 0:
    print("Even")
else:
    print("Odd")

#2 simple calculator
num1 = float(input("Enter the first number: "))
operator = input("Enter Operator( +, -,*,/,%): ")
num2 = float(input("Enter the second number: "))
if operator == "+":
    print("Result", num1 + num2)
elif operator == "-":
    print("Result", num1 - num2)
elif operator == "*":
    print("Result", num1 * num2)
elif operator == "/":
    print("Result", num1 / num2)
elif operator == "%":
    print("Result", num1 % num2)
else:
    print("invalid")

#3 student marks
sub1 = float(input("Enter the Marks of Subjects1: "))
sub2 = float(input("Enter the Marks of Subjects2: "))
sub3 = float(input("Enter the Marks of Subjects3: "))

total = sub1 + sub2 + sub3
percentage = (total / 300) * 100

if percentage >= 50:
    print("Result:Pass")
else:
    print("Result: Fail")

#4 Favorit fruit
fruits = ["apple", "banana", "mango", "orange", "grapes"]
fruit = input("Enter a fruit: ")
if fruit in fruits:
    print("fruit exists in the list:")
else:
    print("Fruit does not exists in the list:")

#5 Student information
student = {
    "name": "Ali",
    "age": 19,
    "favorit_subject": "Computer"
}

print("Name", student["name"])
print("Age", student["age"])
print("Favorit_Subject", student["favorit_subject"])

if student["age"] >= 18:
    print("student is 18 or above")
else:
    print("student is below 18")
```
